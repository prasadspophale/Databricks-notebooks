```python
import pandas as pd

# Create sample data
data = {
    "EmpID": [101, 102, 103],
    "Name": ["Prasad", "Rahul", "Amit"],
    "Salary": [50000, 60000, 70000]
}

# Create DataFrame
df = pd.DataFrame(data)

# Display DataFrame
print(df)
```

```python
df.to_csv("employees.csv", index=False)
print("CSV file created successfully.")
```
